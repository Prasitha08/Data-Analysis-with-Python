## Retail Sales Data Cleaning and Analysis

**Tools:** Google Colab / Jupyter Notebook, Pandas and NumPy  
**Dataset:** `retail_sales_raw.csv` (practice dataset with intentionally missing values)

### Learning objectives
- Import a CSV dataset into Pandas.
- Inspect, clean and organize data.
- Identify and handle missing values.
- Use NumPy and Pandas for basic analysis.
- Create useful summary tables.

> This is a learning exercise using a synthetic practice dataset, not real company data.

```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

# In Google Colab, upload retail_sales_raw.csv using the Files panel.
# In Jupyter, keep the CSV in the same folder as this notebook.
df = pd.read_csv('retail_sales_raw.csv')
df.head()
```

## Step 2: Understand the dataset

```python
print("Shape:", df.shape)
print("\\nColumn names:")
print(df.columns.tolist())
print("\\nData types and non-null counts:")
df.info()
print("\\nStatistical summary:")
df.describe(include='all')
```

## Step 3: Find missing values and duplicates

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\\nDuplicate rows:", df.duplicated().sum())
```

## Step 4: Clean the data

```python
# Convert Date to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Fill missing numeric values with the column median.
numeric_cols = ['Quantity', 'Unit_Price', 'Discount', 'Customer_Rating']
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Remove duplicate rows if any
df = df.drop_duplicates().copy()

# Add calculated sales columns
df['Gross_Sales'] = df['Quantity'] * df['Unit_Price']
df['Net_Sales'] = df['Gross_Sales'] * (1 - df['Discount'])

print("Missing values after cleaning:")
print(df.isnull().sum())
df.head()
```

## Step 5: Use NumPy and Pandas for analysis

```python
# NumPy summary calculations
sales_values = df['Net_Sales'].to_numpy()
print("Mean net sales:", np.mean(sales_values).round(2))
print("Median net sales:", np.median(sales_values).round(2))
print("Standard deviation:", np.std(sales_values).round(2))

# Pandas summaries
print("\\nTotal net sales:", round(df['Net_Sales'].sum(), 2))
print("Average order value:", round(df['Net_Sales'].mean(), 2))
print("\\nSales by category:")
print(df.groupby('Category')['Net_Sales'].sum().sort_values(ascending=False))

print("\\nSales by region:")
print(df.groupby('Region')['Net_Sales'].sum().sort_values(ascending=False))
```

## Step 6: Filter and sort data

```python
# Orders with net sales above 2,000
high_value_orders = df[df['Net_Sales'] > 2000]
display(high_value_orders[['Date','Region','Category','Product','Net_Sales']])

# Top five orders by net sales
display(df.sort_values('Net_Sales', ascending=False).head(5))
```

## Step 7: Export the cleaned dataset

```python
df.to_csv('retail_sales_cleaned.csv', index=False)
print("Cleaned dataset saved as retail_sales_cleaned.csv")
```

## Conclusion
In this module, the raw retail dataset was imported, inspected and cleaned. Missing numeric values were filled using column medians, duplicate rows were removed, and gross/net sales columns were calculated. Pandas group-by operations and NumPy summary functions were used to explore the dataset.
